```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
import os
```

```python
load_dotenv()
```

```python
llm = ChatGoogleGenerativeAI(
    model=os.getenv("GOOGLE_MODEL"),
    api_key=os.getenv("GOOGLE_API_KEY"),
    temperature=0.4,
)
```

```python
class BlogSatate(TypedDict):
    title: str
    outline: str
    blog: str
```

```python
def create_outline(state: BlogSatate) -> BlogSatate:
    title = state['title']
    prompt = f"Create an outline for a blog post with the title: {title}"
    outline = llm.invoke(prompt).content
    state['outline'] = outline
    return state
```

```python
def create_blog(state: BlogSatate) -> BlogSatate:
    title = state['title']
    outline = state['outline']
    prompt = f"Write a blog post with the title: {title} and the following outline: \n {outline}"
    blog_content = llm.invoke(prompt).content
    state['blog'] = blog_content
    return state
```

```python
graph = StateGraph(BlogSatate)

# add node
graph.add_node('create_outline' , create_outline)
graph.add_node('create_blog' , create_blog)

# connect edge
graph.add_edge(START , 'create_outline')
graph.add_edge('create_outline' , 'create_blog')
graph.add_edge('create_outline' , END)

workflow =graph.compile()
```

```python
initial_state = {'title':"rise of the ai in bangladesh"}
final_state = workflow.invoke(initial_state)
print(final_state)
```
